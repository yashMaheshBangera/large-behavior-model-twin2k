```python
def parse_extract_questions(df):
  num_rows = len(df)
  parsed_questions = []
  for i in range(0,num_rows,1):
    current_pid = df["pid"].iloc[i]
    current_persona = df["persona_json"].iloc[i]
    current_persona = json.loads(current_persona)
    current_persona_df = pd.DataFrame(current_persona)
    len_current_persona_df = len(current_persona_df)
    wave = 1
    for j in range(0,len_current_persona_df,1):
      current_question = current_persona_df["Questions"].iloc[j]
      name = current_persona_df["BlockName"].iloc[j].strip()
      if name == "Personality" and j > 1:    # 2nd and 3rd Personality blocks
          wave += 1
      if name == "False consensus":
          wave = 4
      for k in range(0,len(current_question),1):
        q = current_persona_df["Questions"].iloc[j][k]
        current_dictionary = {
            "pid":current_pid,
            "element":j,
            "block_name":name,
            "block_type":current_persona_df["BlockType"].iloc[j],
            "wave":wave,
            "qID":q["QuestionID"],
            "qtext":q["QuestionText"],
            "qtype":q["QuestionType"],
            "options":q.get("Options"),
            "settings_selector":q.get("Settings",{}).get("Selector"),
            "settings_subSelector":q.get("Settings",{}).get("SubSelector"),
            "settings_forceResponse":q.get("Settings",{}).get("ForceResponse"),
            "answers_selected_position":q.get("Answers",{}).get("SelectedByPosition"),
            "answers_selected_text":q.get("Answers",{}).get("SelectedText")
        }
        parsed_questions.append(current_dictionary)
  return pd.DataFrame(parsed_questions)
```

```python
def extract_raw_answers(df):
    rows = []
    for pid, pj in zip(df["pid"], df["persona_json"]):
        persona = json.loads(pj)
        for element, block in enumerate(persona):
            for q in block.get("Questions", []):
                rows.append({
                    "pid": pid,
                    "element": element,
                    "qID": q["QuestionID"],
                    "answers_raw": q.get("Answers"),
                })
    return pd.DataFrame(rows)

raw_df = extract_raw_answers(fp_df)
```

```python
def extract_matrix_answers(df):
    rows = []
    for pid, pj in zip(df["pid"], df["persona_json"]):
        for element, block in enumerate(json.loads(pj)):
            for q in block.get("Questions", []):
                if q["QuestionType"] != "Matrix":
                    continue
                positions = q["Answers"]["SelectedByPosition"]
                texts = q["Answers"]["SelectedText"]
                if len(positions) != len(q["Rows"]):
                    continue        # rows skipped: alignment unknown, so leave out
                for i, (row_text, pos, txt) in enumerate(zip(q["Rows"], positions, texts), start=1):
                    rows.append({
                        "pid": pid,
                        "element": element,
                        "qID": q["QuestionID"],
                        "sub_item": i,
                        "sub_item_text": row_text,
                        "answer_pos": pos,
                        "answer_text": txt,
                        "n_scale": len(q["Columns"]),
                    })
    return pd.DataFrame(rows)

matrix_df = extract_matrix_answers(fp_df)
```
