```python
import json
```

```python
def check_unique_list(df,block_names,count):
  fp_persona_sample = json.loads(df["persona_json"][count])
  current_block_names = []
  for i in fp_persona_sample:
    current_block_names.append(i["BlockName"])
  return current_block_names
```

```python
block_names = []
for count in range(0,len(fp_df),1):
  current_block_names = check_unique_list(fp_df,block_names,count)
  block_names.append(current_block_names)
```

```python
def parse_extract_questions(df):
  num_rows = len(df)
  parsed_questions = []
  for i in range(0,num_rows,1):
    current_pid = df["pid"].iloc[i]
    current_persona = df["persona_json"].iloc[i]
    current_persona = json.loads(current_persona)
    current_persona_df = pd.DataFrame(current_persona)
    len_current_persona_df = len(current_persona_df)
    print(len_current_persona_df)
    for k in range(0,len_current_persona_df,1):
        current_dictionary = {
            "pid":current_pid,
            "qID":current_persona_df["Questions"].iloc[k][0]["QuestionID"],
            "qtext":current_persona_df["Questions"].iloc[k][0]["QuestionText"],
            "qtype":current_persona_df["Questions"].iloc[k][0]["QuestionType"],
            #"options":current_persona_df["Questions"].iloc[k][0]["Options"],
            "settings_selector":current_persona_df["Questions"].iloc[k][0]["Settings"]["Selector"],
            "settings_subSelector":current_persona_df["Questions"].iloc[k][0]["Settings"]["SubSelector"],
            "settings_forceResponse":current_persona_df["Questions"].iloc[k][0]["Settings"]["ForceResponse"],
            "answers_selected_position":current_persona_df["Questions"].iloc[k][0]["Answers"]["SelectedByPosition"],
            "answers_selected_text":current_persona_df["Questions"].iloc[k][0]["Answers"]["SelectedText"]
        }
        parsed_questions.append(pd.DataFrame(current_dictionary))
  return parsed_questions
```
